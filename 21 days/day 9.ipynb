#### Q1. Take a number and print all its factors.

```
n=int(input("Enter a number: "))

for i in range(1, n+1):
    if n % i == 0:
        print(i, end=" ")
```

#### Q2. Take a list and count how many prime numbers are present.

```
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

lst=[int(x) for x in input("Enter numbers with space: ").split()]
primes = [x for x in lst if is_prime(x)]
print("Prime numbers in the list:", primes)
```

#### Q3. Take a string and print characters that appear more than once.

```
s=input("Enter a string: ")
ferq={}
for c in s:
    ferq[c] = ferq.get(c, 0) + 1

for k, v in ferq.items():
    if v > 1:
        print(f"{k}: {v}")
```

#### Q5. Take a sentence and print longest word.

```
s=input("Enter a sentence: ")
words = s.split()
longest = max(words, key=len)
print("Longest word:", longest)
```

#### Q8. Remove duplicate characters from string.

```
s=input("Enter a string: ")
result=""
for c in s:
    if c not in result:
        result+=c
print(result)
```

#### Q9. Find sum of digits using recursion.

```
def sum_of_digits(n):
    if n == 0:
        return 0
    else:
        return n % 10 + sum_of_digits(n // 10)

n = int(input("Enter a number: "))
print("Sum of digits:", sum_of_digits(n))
```

#### Q10. Count how many elements are greater than average.

```
lst = [1, 2, 3, 4, 5]
avg = sum(lst) / len(lst)
count = sum(1 for x in lst if x > avg)
print("Number of elements greater than average:", count)
```

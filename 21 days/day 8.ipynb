#### Q1. Take a string and print frequency of vowels using dictionary.

```
s=input("Enter a string: ").lower()
freq={"a":0,"e":0,"i":0,"o":0,"u":0}

for ch in s:
    if ch in freq:
        freq[ch] += 1

print(freq)
```

#### Q2. Take a list and return only prime numbers using function.

```
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def get_primes(lst):
    result = []
    for x in lst:
        if is_prime(x):
            result.append(x)
    return result

print(get_primes([1,2,3,4,5,6,7,8]))
```

#### Q3. Count words in a sentence and store in dictionary.

```
s=input("Enter a sentence: ")
words=s.split()
freq={}
for w in words:
    freq[w] = freq.get(w, 0) + 1
print(freq)
```

#### Q4. Take list and print sum of squares using loop.

```
lst=[int(x) for x in input("Enter numbers with space: ").split()]

toal=0
for x in lst:
    toal+=x*x

print("Total:", toal)
```

#### Q6. Reverse each word in a sentence.

```
s = input("Enter sentence: ")

words = s.split()

for w in words:
    print(w[::-1], end=" ")
```

#### Q7. Find largest number in list using function (no max).

```
def largest_list(lst):
    if not lst:
        return None
    max_val = lst[0]
    for x in lst:
        if x > max_val:
            max_val = x
    return max_val

x=[int(x) for x in input("Enter numbers with space: ").split()]
print("Largest number:", largest_list(x))
```

#### Q9. Count uppercase and lowercase letters.

```
s=input("Enter a string: ")

upper=lower=0
for c in s:
    if c.isupper():
        upper+=1
    elif c.islower():
        lower+=1
print(f"Uppercase letters: {upper}")
print(f"Lowercase letters: {lower}")
```

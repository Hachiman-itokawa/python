#### Q8. Remove duplicate characters from string.


```python
s=input("Enter a string: ")
result=""
for c in s:
    if c not in result:
        result+=c
print(result)
```

#### 2 Check whether a number is an Armstrong number.


```python
n = int(input("Enter a number: "))
temp = n
s = 0
while temp > 0:
    digit = temp % 10
    s += digit ** 3
    temp //= 10
if s == n:
    print(f"{n} is an Armstrong number.")
else:
    print(f"{n} is not an Armstrong number.")
```

#### 3. Check palindrome string.


```python
m=input("Enter a string: ")
if m==m[::-1]:
    print(f"{m} is a palindrome string.")
else:
    print(f"{m} is not a palindrome string .")
```

#### 7. Write a function to find factorial.


```python

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
n = int(input("Enter a number: "))
print("Factorial =", factorial(n))
```

#### 8. Take a list and print sum of squares using loop.


```python
# +code
lst = [1, 2, 3, 4, 5]
total = 0
for x in lst:
    total += x * x
print("Sum of squares:", total)
```

#### 9 Take a string and print characters that appear more than once.


```python

s = input("Enter a string: " )
freq = {}
for c in s:
    freq[c] = freq.get(c, 0) + 1
for ch, count in freq.items():
    if count > 1:
        print(f"{ch}: {count}")
```

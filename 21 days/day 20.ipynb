#### 1. Write a lambda function to find maximum of two numbers.

```python
maxi=lambda a,b: a if a>b else b
print(maxi(int(input("Enter first number: ")), int(input("Enter second number: "))))
```

#### 2. Count words in a sentence and store in dictionary.

```python
s=input("Enter a sentence: ")
words=s.split()
freq={}
for w in words:
    freq[w] = freq.get(w, 0) + 1
print(freq)
```

#### 3. Take a number and print all its factors.

```python
n=int(input("Enter a number: "))

for i in range(1, n+1):
    if n % i == 0:
        print(i, end=" ")
```

#### 4. Find all pairs in list with given sum.

```python
lst=[1,2,3,4,5,6,7,8,9]
target=int(input("Enter target sum: "))
for i in range(len(lst)):
    for j in range(i+1,len(lst)):
        if lst[i]+lst[j]==target:
            print(f"Pair: ({lst[i]}, {lst[j]})")
```

#### 5 Check whether a number is an Armstrong number.


```python
n = int(input("Enter a number: "))
temp = n
s = 0
while temp > 0:
    digit = temp % 10
    s += digit ** 3
    temp //= 10
if s == n:
    print(f"{n} is an Armstrong number.")
else:
    print(f"{n} is not an Armstrong number.")
```

#### 6. Find average of elements in a tuple.


```python
t = (10, 20, 30, 40)
avg = sum(t) / len(t)
print("Average =", avg)
```

#### 8. Write a recursive function for Fibonacci.

```python
def fib(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b

n=int(input("Enter a number: "))
fib(n)
```

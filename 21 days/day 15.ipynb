#### 7. Write a function to find maximum of two numbers.


```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Maximum =", maximum(a, b))
```

#### 8. Take a string and print frequency of vowels using dictionary.


```python
s = input("Enter a string: " ).lower()
freq = {"a":0, "e":0, "i":0, "o":0, "u":0}
for ch in s:
    if ch in freq:
        freq[ch] += 1
print(freq)
```

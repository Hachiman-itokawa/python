#### Q1. Take a string and print each character using loop.

```
s=input("Enter a string: ")
for i in s:
    print(i)
```

#### Q2. Print string with index using enumerate().

```
s=input("Enter a string: ")
for i,c in enumerate(s):
    print(f"Index: {i}, Character: {c}")
```

#### Q6. Check palindrome string.

```
m=input("Enter a string: ")
if m==m[::-1]:
    print(f"{m} is a palindrome string.")
else:
    print(f"{m} is not a palindrome string .")
```

#### Q7. Count vowels in a string.

```
s=input("Enter a string: ")
vowels="aeiouAEIOU"
count=0
for c in s:
    if c in vowels:
        count+=1
print(f"The number of vowels in the string is: {count}")
```

#### Q8. Count uppercase, lowercase and digits.

```
s=input("Enter a string: ")
upper=0
lower=0
digits=0
for c in s:
    if c.isupper():
        upper+=1
    elif c.islower():
        lower+=1
    elif c.isdigit():
        digits+=1
print(f"The number of uppercase letters in the string is: {upper}")
print(f"The number of lowercase letters in the string is: {lower}")
print(f"The number of digits in the string is: {digits}")
```

#### Q13. Find frequency of each character.

```
s = input("Enter a string: ")

freq = {}

for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
```

#### Q17. Count frequency of words in sentence.

```
s = input("Enter a sentence: ").lower()

words = s.split()
freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1

print(freq)
```

#### Q18. Find longest word in sentence.

```
s = input("Enter a sentence: ")

words = s.split()
longest = max(words, key=len)

print("Longest word:", longest)
```

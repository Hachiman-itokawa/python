#### Q1. Take a list and print all elements using loop.

```
List=[10, 20, 30, 40]
for x in List:
    print(x)
```

#### Q2. Find sum of elements in a list.

```
total=0
for x in List:
    total+=x    
print("Total:",total)
```

#### Q3. Find maximum and minimum element.

```
print("maximum =",max(List))
print("minimum =",min(List))
```

#### Q4. Count frequency of each element.

```
List=[1,2,3,1,5,4,3,2,4,1]
freq={}
for x in List:
    freq[x] = freq.get(x, 0) + 1
print(freq)
```

#### Q5. Remove duplicates using set.

```
unique=list(set(List))
print("Unique elements:", unique)
```

#### Q6. Find second largest number.

```
List.sort()
print("second largest:", List[-2])
```

#### Q7. Merge two lists.

```
l2=[5, 10, 15, 20]
merged=unique+l2
print("Merged list:", merged)
```

#### Q8. Find common elements between two lists.

```
common=[]

for x in unique:
    if x in l2:
        common.append(x)
print("Common elements:", common)
```

#### Q9. Separate even and odd numbers.

```
even=[]
odd=[]

for x in merged:
    if x%2==0:
        even.append(x)
    else:
        odd.append(x)

print("Even numbers:", even)
print("Odd numbers:", odd)
```

#### Q10. Sort list without using sort().

```
newl=[1,21,323,13,12,42,12,14]
for i in range(len(newl)):
    for j in range(i+1,len(newl)):
        if newl[i]>newl[j]:
            newl[i],newl[j]=newl[j],newl[i]
print( newl)
```

#### Q11. Create list of squares using list comprehension.

```
squares = [x*x for x in range(1, 6)]
print(squares)
```

#### Q12. Filter even numbers using list comprehension.

```
even=[x for x in l2 if x%2==0]
print( even)
```

#### Q13. Reverse a list without using built-in function.

```
rev=[]

for i in  range(len(newl)-1,-1,-1):
    rev.append(newl[i])
print( rev)
```

#### Q14. Remove all occurrences of a given element.

```
result=[]
for i in merged:
    if i%2!=0:
        result.append(i)
print(result)
```

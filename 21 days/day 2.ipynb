#### Q3. Find factorial of a number.

```
n=int(input("Enter a number: "))
fact=1
for i in range(1,n+1):
    fact*=i
print(f"Factorial of {n} is {fact}")
```

#### Q4. Check whether a number is Armstrong number.

```
n=int(input("Enter a number: "))
temp=n
sum=0
while temp>0:
    digit=temp%10
    sum+=digit**3
    temp//=10
if sum==n:
    print(f"{n} is an Armstrong number.")
else:
    print(f"{n} is not an Armstrong number.")
```

#### Q6. Check whether a number is prime.

```
n=int(input("Enter a number: "))
if n>1:
    for i in range(2,n):
        if n%i==0:
            print(f"{n} is not a prime number.")
            break
    else:
        print(f"{n} is a prime number.")
```

#### Q7. Check palindrome number.

```
n=int(input("Enter a number: "))
temp=n
rev=0
while temp>0:
    digit=temp%10
    rev=rev*10+digit
    temp//=10
if rev==n:
    print(f"{n} is a palindrome number.")
else:
    print(f"{n} is not a palindrome number.")
```

#### Q8. Print sum of digits of a number.

```
n=int(input("Enter a number: "))
sum=0
for i in range(1,n+1):
    sum+=i
print(f"Sum of first {n} natural numbers is {sum}.")
```

#### Q9. Print numbers divisible by 5 or 7 between 1 to 100.

```
for i in range (1,101):
    if i%5==0 or i%7==0:
        print(i,end=" ")
```

#### Q10. Print all prime numbers between 1 to 50.

```
for num in range(1,51):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num,end=" ")
```

#### Q11. Demonstrate break statement.

```
for i in range(10):
    if i == 5:
        break
    print(i)
```

#### Q12. Demonstrate continue statement.

```
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

#### Q16. Loop through dictionary (key-value pairs).

```
d = {"A": 90, "B": 85, "C": 92}

for k, v in d.items():
    print(k, v)
```

#### Q17. Create list of squares using loop.

```
squares = []

for i in range(1, 6):
    squares.append(i*i)

print(squares)
```

#### Q18. Same using list comprehension.

```
squares = [x*x for x in range(1, 6)]
print(squares)
```

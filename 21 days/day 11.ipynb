#### Q1. Student Marks System (Dictionary)
Take 3 studentsâ€™ marks and print average marks.

```python

marks = {}
for i in range(3):
    name = input("Enter student name: ")
    mark = float(input("Enter marks: "))
    marks[name] = mark

avg = sum(marks.values()) / len(marks)
print("Average marks: %.2f" %avg)

```

#### Q2. Menu-driven Calculator

```python
while True:
    print("Menu:")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")
    
    choice = input("Enter your choice: ")
    
    if choice == '5':
        print("Exiting...")
        break
    
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    
    if choice == '1':
        print(f"Result: {num1 + num2}")
    elif choice == '2':
        print(f"Result: {num1 - num2}")
    elif choice == '3':
        print(f"Result: {num1 * num2}")
    elif choice == '4':
        if num2 != 0:
            print(f"Result: {num1 / num2}")
        else:
            print("Error: Division by zero")
    else:
        print("Invalid choice, please try again.")
```

#### Q6. Find second largest in list (no built-in sort)

```python
lst=[10,20,5,15,25]
largest=second_largest=float('-inf')
for x in lst:
    if x>largest:
        second_largest=largest
        largest=x
    elif largest>x>second_largest:
        second_largest=x
print("Second largest:", second_largest)
```

#### Q7. Check strong number

(sum of factorial of digits)

```python
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
n=int(input("Enter a number: "))
temp=n
sum=0
while temp>0:
    digit=temp%10
    sum+=factorial(digit)
    temp//=10  
if sum==n:
    print(f"{n} is a strong number.")
```

#### Q8. Count duplicate words in sentence

```python

sentence = "This is a test sentence with some duplicate words. This sentence is a test."
words = sentence.split()
duplicate_words = []
for word in words:
    if words.count(word) > 1 and word not in duplicate_words:
        duplicate_words.append(word)
print("Duplicate words:", duplicate_words)
```

#### Q10. Find all Armstrong numbers between 1–500

```python


armstrong_numbers = []
for num in range(1, 501):
    order = len(str(num))
    sum = 0
    temp = num
    while temp > 0:
        digit = temp % 10
        sum += digit ** order
        temp //= 10
    if num == sum:
        armstrong_numbers.append(num)

print("Armstrong numbers between 1 and 500:", armstrong_numbers)
```

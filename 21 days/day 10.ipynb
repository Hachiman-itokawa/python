#### Q1. Check if a number is perfect number.

```
n = int(input("Enter a number: "))
divisors = [i for i in range(1, n) if n % i == 0]
if sum(divisors) == n:
    print(f"{n} is a perfect number.")
else:
    print(f"{n} is not a perfect number.")
```

#### Q2. Find intersection of two lists without using set.

```
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
intersection = [x for x in list1 if x in list2]
print("Intersection:", intersection)
```

#### Q3. Check if two strings are anagrams.

```
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
if sorted(s1) == sorted(s2):
    print("Anagrams")
else:
    print("Not anagrams")
```

#### Q4. Find all pairs in list with given sum.

```
lst=[1,2,3,4,5,6,7,8,9]
target=int(input("Enter target sum: "))
for i in range(len(lst)):
    for j in range(i+1,len(lst)):
        if lst[i]+lst[j]==target:
            print(f"Pair: ({lst[i]}, {lst[j]})")
```

#### Q5. Print Pascal-like pattern (simple version).

```
n=int(input("Enter number of rows: "))
for i in range(1, n+1):
    print(['*']*i)
```

#### Q8. Find most frequent element in list.

```
#Q8. Find most frequent element in list.
lst=[1,2,3,4,5,2,3,4,2]
freq={}
for x in lst:
    freq[x] = freq.get(x, 0) + 1
max_value=max(freq.values())
print(max_value)
```

#### Q9. Flatten a 2D list.

```
lst=[[1,2],[3,4],[5,6]]
flat=[x for sublist in lst for x in sublist]
print(flat)
```

#### Q10. Count words starting with vowel.

```
s=input("Enter a sentence: ").lower()
words=s.split()
count=0
for w in words:
    if w[0] in "aeiou":
        count+=1
print("Number of words starting with vowel:", count)
```

#### Q1. Create a dictionary of student (name & marks) and print it.

```
student = {"Rahul": 85, "Aman": 90, "Neha": 88}

print(student)
```

#### Q2. Print all keys and values separately.

```
d = {"A": 10, "B": 20, "C": 30}

print("Keys:", d.keys())
print("Values:", d.values())
```

#### Q3. Print key-value pairs using loop.

```
d = {"A": 10, "B": 20, "C": 30}

for k, v in d.items():
    print(k, v)
```

#### Q4. Add a new key-value pair to dictionary.

```
d = {"A": 10, "B": 20}

d["C"] = 30

print(d)
```

#### Q5. Update value of a key.

```
d = {"A": 10, "B": 20}

d["A"] = 50

print(d)
```

#### Q6. Find key with maximum value.

```
d = {"A": 10, "B": 50, "C": 30}

max_key = max(d, key=d.get)

print("Max key:", max_key)
```

#### Q7. Check if a key exists in dictionary.

```
d = {"A": 10, "B": 20}

key = input("Enter key: ")

if key in d:
    print("Exists")
else:
    print("Not Exists")
```

#### Q8.Create a tuple and print its elements.

```
t = (10, 20, 30)

for x in t:
    print(x)
```

#### Q9. Find average of elements in tuple.

```
t = (10, 20, 30, 40)

avg = sum(t) / len(t)

print("Average = %d" %avg)
```

#### Q10. Multiply all elements of list.

```
lst = [1, 2, 3, 4]

product = 1

for x in lst:
    product *= x

print("Product =", product)
```

#### Q11. Insert element at specific position.

```
lst = [1, 2, 4, 5]

pos = int(input("Enter position: "))
val = int(input("Enter value: "))

lst.insert(pos, val)

print(lst)
```

#### Q1. Write a function to find maximum of two numbers.

```
def maximum(a, b):
    if a > b:
        return a
    else:
        return b
a=int(input("Enter first number: "))
b=int(input("Enter second number: "))
print("Maximum =", maximum(a, b))
```

#### Q2. Write a function to find factorial.

```
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
n=int(input("Enter a number: "))
print("Factorial =", factorial(n))
```

#### Q3. Write a function to check prime number.

```
def prime(n):
    if n>1:
        for i in range(2,n):
            if n%i==0:
                return False
        else:
            return True
    else:
        return False
n=int(input("Enter a number: "))
print(f"is {n} prime no:", prime(n))
```

#### Q4. Write a function to find sum of digits.

```
def sum_of_digits(n):
    sum=0
    while n>0:
        digit=n%10
        sum+=digit
        n//=10
    return sum
n=int(input("Enter a number: "))
print("Sum of digits =", sum_of_digits(n))
```

#### Q5. Write a function to return multiple values (sum & average).

```
def sum_and_avearge(lst):
    total=sum(lst)
    average=total/len(lst)
    return total, average

lst=[int(x) for x in input("Enter numbers with space: ").split()]
total, average = sum_and_avearge(lst)
print("Total =", total)
print("Average =", average)
```

#### Q6. Write a recursive function for Fibonacci.

```
def fib(n):
    a, b = 0, 1
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b

n=int(input("Enter a number: "))
fib(n)
```

#### Q9. Write a function to reverse a number.

```
def reverse_no(n):
    rev=0
    while n>0:
        digit=n%10
        rev=rev*10+digit
        n//=10
    return rev

n=int(input("Enter a number: "))
print("Reverse =", reverse_no(n))
```

#### Q10. Write a function to count vowels in a string.

```
def count_vowels(s):
    count=0
    for c in s.lower():
        if c in "aeiou":
            count+=1
    return count

s=input("Enter a string: ")
print("Number of vowels =", count_vowels(s))
```

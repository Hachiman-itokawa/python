<h4>Q1. Print pattern:</h4>

```python
for i in range(1, 5):
    print("*" * i)
```

<h4>Q2. Print pattern:</h4>

```python
for i in range(1, 5):
    for j in range(1, i+1):
        print(j, end="")
    print()
```

<h4>Q3. Print reverse pattern:</h4>
<h5>****<br>
***<br>
**<br>
*</h5>

```python
for i in range(4, 0, -1):
    print("*" * i)
```

<h4>Q4. Print pyramid:</h4>

```python
n = 4
for i in range(1, n+1):
    print(" "*(n-i) + "*"*(2*i-1))
```

<h4>Q5. Print numbers in triangle form:</h4>
<h5>1<br>
23<br>
456<br>
78910</h5>

```python
num = 1
for i in range(1, 5):
    for j in range(i):
        print(num, end="")
        num += 1
    print()
```

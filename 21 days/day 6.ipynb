#### Q1. Create a set and print all elements.

```
s = {10, 20, 30, 40}

for x in s:
    print(x)
```

#### Q2. Add elements to a set.


```
n=int(input("Enter a number: "))
s.add(n)
print(s)
```

#### Q3. Remove an element from set.

```
n=int(input("Enter a number: "))
s.remove(n)
print(s)
```

#### Q4. Use discard() to remove element safely.

```
s.discard(5)  # no error even if not present

print(s)
```

#### Q5. Find union of two sets.

```
A = {1, 2, 3}
B = {3, 4, 5}

A.union(B)
```

#### Q6. Find intersection of two sets.

```
A.intersection(B)
```

#### Q7. Find difference of two sets.

```
print(A - B)
```

#### Q8. Check if element exists in set.

```
s = {10, 20, 30}

x = int(input("Enter one digit: "))

if x*10 in s:
    print("Exists")
else:
    print("Not Exists")
```

#### Q9. Convert list into set to remove duplicates.

```
lst = [1, 2, 2, 3, 4, 4]

unique = set(lst)

print(unique)
```

#### Q10. Find symmetric difference between two sets.

```
A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
```

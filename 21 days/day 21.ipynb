#### 2. Take a list and count how many prime numbers are present.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

lst=[int(x) for x in input("Enter numbers with space: ").split()]
primes = [x for x in lst if is_prime(x)]
print("Prime numbers in the list:", primes)
```

#### 4. Student Marks System (Dictionary)
Take 3 students marks and print average marks.

```python
marks = {}
for i in range(3):
    name = input("Enter student name: ")
    mark = float(input("Enter marks: "))
    marks[name] = mark

avg = sum(marks.values()) / len(marks)
print("Average marks: %.2f" %avg)
```

#### 9. WAP to enter a string and a substring. You have to print the number of times that the substring occurs in the given string. String traversal will take place from left to right, not from right to left.

```python
main_string=input("enter a string:")#banana
sub_string=input("enter a sub string:")#na
count=0
n=len(main_string)#6
m=len(sub_string)#2
for i in range(n-m+1):
    if(main_string[i:i+m]==sub_string):
        count+=1

print("",count)
```

#### 10. Scan n values in range 0-3 and print the number of times each value has occurred.

```python
n=int(input("enter a number: "))
freq=[0,0,0,0]

for i in range(n):
    num=int(input("enter a number(0-3): "))
    if 0<=num<=3:
        freq[num]+=1
for i in range(4):
    print(i,"occurred", freq[i],"times")
```
